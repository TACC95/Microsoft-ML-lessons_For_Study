```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from sklearn import datasets, linear_model, model_selection
```

```python
X, y = datasets.load_diabetes(return_X_y=True)
X
```

```python
X = X[:, 2] # 第三列数据是bmi数据
```

```python
X = X.reshape((-1, 1))
X.shape
```

```python
X_train, X_test, y_train, y_test = model_selection.train_test_split(X, y, test_size=0.33)
```

```python
model = linear_model.LinearRegression()
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
plt.scatter(X_test, y_test, color='black')
plt.plot(X_test, y_pred, color='b', linewidth=3)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from sklearn import datasets, linear_model, model_selection
import seaborn as sns
```

```python
lin = datasets.load_linnerud()
print(lin.data.shape)
print(lin.target.shape)
```

```python
print(lin.feature_names)
print(lin.target_names)
```

可以看出，在这个数据集中，特征数据集与运动类型有关，目标值数据集与生理数据有关。作业的要求是要分析腰围与仰卧起坐之间的关系。这里我们使用线性回归，然后用seaborn的多项式回归。

```python
X = lin.data
X = X[:, 1]
```

```python
X = X.reshape((-1, 1))
X.shape
```

```python
y = lin.target
```

```python
y = y[:, 1]
y
```

```python
X_train, X_test, y_train, y_test = model_selection.train_test_split(X, y, test_size=0.33)
```

```python
model = linear_model.LinearRegression()
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
plt.scatter(X_test, y_test)
plt.plot(X_test, y_pred)
plt.show()
```

可以看到，腰围与仰卧起坐呈负相关，即腰围越大，能做的仰卧起坐个数就越少。

## seaborn的多项式回归

```python
sns.regplot(x=X_train, y=y_train, order=3)
plt.show()
```

用三次多项式进行回归分析，仍然呈负相关。
